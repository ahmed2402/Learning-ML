### DBSCAN Clustering:

DBSCAN (Density-Based Spatial Clustering of Applications with Noise) is a clustering algorithm that groups together data points that are close to each other based on a distance measurement and a minimum number of points.

#### How does it work?
- DBSCAN looks for areas where data points are packed closely together (dense regions).
- It starts with a random point and checks how many points are within a certain distance (called `eps`).
- If there are enough points nearby (at least `min_samples`), it forms a cluster.
- It then expands the cluster by checking the neighbors of neighbors, and so on.
- Points that do not belong to any cluster (because they are too far from others) are labeled as noise or outliers.

#### Basic Terminologies:
- **Core Point:** A point with at least `min_samples` points (including itself) within distance `eps`.
- **Border Point:** A point that is within `eps` of a core point but does not have enough neighbors to be a core point itself.
- **Noise Point (Outlier):** A point that is not a core point and not close enough to any core point to be included in a cluster.
- **eps (epsilon):** The maximum distance between two points for them to be considered neighbors.
- **min_samples:** The minimum number of points required to form a dense region (a cluster).

DBSCAN is useful because it can find clusters of different shapes and sizes, and it can also identify outliers in the data.


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_moons
from sklearn.cluster import DBSCAN
```

```python
X,y = make_moons(n_samples=250, noise=0.05)
```

```python
plt.scatter(X[:,0],X[:,1])
```

```python
from sklearn.preprocessing import StandardScaler
se = StandardScaler()
X_scaled = se.fit_transform(X)
```

```python
from sklearn.cluster import DBSCAN
db = DBSCAN(eps=0.5)

db.fit(X_scaled)

plt.scatter(X_scaled[:,0],X_scaled[:,1],c=db.labels_)
plt.show()
```
